KPI1: Gross Merchandise Value (GMV)

```python
gmv = orders["final_amount"].sum()
print("Gross Merchandise Value for all Status:", gmv)
```

Business meaning: This represents the total value of orders in the dataset, including orders that may not have been delivered.

KPI2: Net Revenue

```python
net_revenue = orders.loc[orders["status"] == "Delivered","final_amount"].sum()
print("Net Revenue:",round(net_revenue,2))
```

Business meaning: This shows the value associated with completed deliveries and helps distinguish delivered business from other order statuses.

KPI3: Average Order Value (AOV)

```python
delivered_orders = orders.loc[orders["status"] == "Delivered", "order_id"].count()


aov = net_revenue/delivered_orders

print("Delivered Orders:", delivered_orders)
print("Average Order Value:",round(aov,2))
```

Business Insight: AOV indicates the average revenue generated per delivered order. Monitoring AOV helps the business understand customer spending patterns and identify opportunities to increase order value through product bundling, cross-selling, and targeted promotions.

KPI4: Cancellation Rate

```python
total_orders = orders["order_id"].count()
cancelled_orders = orders.loc[orders["status"] == "Cancelled","order_id"].count()

cancellation_rate = (cancelled_orders/total_orders) * 100

print("Total Orders:",total_orders)
print("Cancelled Orders:",cancelled_orders)
print("Cancellation Rate:",round(cancellation_rate),"%")
```

Business meaning: Cancelled orders may represent lost sales and wasted operational effort. The main cancellation reasons need to be investigated.

KPI5: Return Rate

```python
return_records = returns["return_id"].count()

return_rate = (return_records/delivered_orders) * 100

print("Total Return Records:",return_records)
print("Return Rate:",round(return_rate),"%")
```

Business meaning: Returns may increase reverse-shipping and processing costs and affect customer satisfaction. The final rate and product-level causes should be verified before submission.

KPI6: Customer Lifetime Value (CLV)

```python
clv = customers.groupby("segment")["total_spent"].mean()
print("Customer Lifetime Value:",round(clv,2))
```

Business meaning: Premium customers may be a valuable segment to retain. Further analysis of repeat purchases and profit would provide a fuller picture of customer value.

KPI7: Month-over-Month Growth

```python
orders["order_date"] = pd.to_datetime(orders["order_date"])

orders["month"] = orders["order_date"].dt.to_period("M")

monthly_gmv = orders.groupby("month")["final_amount"].sum()

print(monthly_gmv)
```

```python
mom_growth = monthly_gmv.pct_change() * 100

print(mom_growth.round(2))
```

Business Insight: Month-over-month GMV fluctuates throughout the period, with a sharp increase of 370.77% in July 2023 and the largest decline of 23.74% in June 2025. The business should investigate the reasons behind these changes and use monthly sales trends to improve demand forecasting, marketing campaigns, and inventory planning.

Year-over-Year Growth 

```python
orders["year"] = orders["order_date"].dt.year

yearly_gmv = orders.groupby("year")["final_amount"].sum()

print(yearly_gmv)
```

```python
yoy_growth = yearly_gmv.pct_change() * 100

print(yoy_growth.round(2))
```

Business Insight: GMV grew by 88.28% in 2024, indicating strong business growth, but declined by 51.36% in 2025. The business should investigate the reasons for the decline and review sales trends, order volume, and customer demand to identify corrective actions. First, confirm whether 2025 data covers the full year.

KPI8: Top Category Revenue Share

```python
category_revenue = order_items.groupby("category")["total_price"].sum()
total_revenue = category_revenue.sum()

revenue_share = (category_revenue / total_revenue) * 100

print("Category Revenue",category_revenue)
print("Revenue Share",revenue_share.round(2),)
```

Business meaning: Electronics is a major revenue driver, but its large share creates some dependence on one category. Growing other categories may help diversify revenue.

KPI9: Payment Failure Rate

```python
failed_payments = payments.loc[payments["status"] == "Failed","payment_id"].count()
total_payments = payments["payment_id"].count()

payment_failure_rate = (failed_payments/total_payments) * 100

print("Total Failed Payments",failed_payments)
print("Total Payments",total_payments)
print("Payment Failure Rate",round(payment_failure_rate),"%")
```

Target: Below 2%    
Business meaning: Some customers may be unable to complete their purchases. IndiaKart should investigate failure rates by payment method.

KPI10: Inventory Fill Rate

```python
in_stock_skus = inventory.loc[inventory["quantity_available"] > 0, "product_id"].nunique()

total_skus = products["sku"].nunique()

inventory_fill_rate = (in_stock_skus / total_skus) * 100

print("In-Stock Products:", in_stock_skus)
print("Total SKUs:", total_skus)
print("Inventory Fill Rate:", round(inventory_fill_rate, 2), "%")
```

Target > 95%.   
Business meaning: Most tracked products appear to be in stock. IndiaKart should maintain availability while monitoring low-stock products and replenishment needs.

Bonus: Customer Cohort Analysis

```python
print(orders.columns)
print(customers.columns)
```

```python
customers["registration_date"] = pd.to_datetime(
    customers["registration_date"], dayfirst=True
)

orders["order_date"] = pd.to_datetime(
    orders["order_date"], dayfirst=True
)
```

```python
customers["cohort_month"] = (
    customers["registration_date"].dt.to_period("M")
)

print(customers[["customer_id", "registration_date", "cohort_month"]].head())
```

```python
cohort_customers = customers.groupby("cohort_month")["customer_id"].nunique()

print(cohort_customers)
```

```python
cohort_data = orders.merge(
    customers[["customer_id", "cohort_month"]],
    on="customer_id",
    how="left"
)

print(cohort_data[
    ["customer_id", "cohort_month", "order_date", "order_id"]
].head())
```

```python
cohort_data = orders.merge(
    customers[["customer_id", "cohort_month"]],
    on="customer_id",
    how="left"
)

print(cohort_data[
    ["customer_id", "cohort_month", "order_date", "order_id"]
].head())
```

```python
cohort_data["order_month"] = (
    cohort_data["order_date"].dt.to_period("M")
)

cohort_data["cohort_index"] = (
    (cohort_data["order_month"].dt.year
     - cohort_data["cohort_month"].dt.year) * 12
    + (cohort_data["order_month"].dt.month
       - cohort_data["cohort_month"].dt.month)
    + 1
)
```

```python
cohort_data = cohort_data[
    cohort_data["cohort_index"] >= 1
].copy()

print("Valid cohort order records:", len(cohort_data))
```

```python
cohort_table = cohort_data[
    cohort_data["cohort_index"].isin([1, 2, 3])
].pivot_table(
    index="cohort_month",
    columns="cohort_index",
    values="order_id",
    aggfunc="nunique",
    fill_value=0
)

cohort_table.columns = ["Month 1", "Month 2", "Month 3"]

print(cohort_table)
```

Business meaning: This helps IndiaKart examine how customer activity changes after registration and identify opportunities to encourage repeat purchases.
