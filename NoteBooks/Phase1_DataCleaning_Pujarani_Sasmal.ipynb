IndiaCart E-Commerce Analytics
Phase 1: Data Cleaning Summary
Objective

The objective of Phase 1 was to inspect and clean the IndiaCart e-commerce datasets to improve data quality and prepare the data for Exploratory Data Analysis (EDA), KPI analysis, and dashboard development.

Data Quality Checks Performed

The datasets were loaded into Python and systematically checked for:

Missing/null values
Incorrect data types
Duplicate records
Statistical outliers in relevant numerical columns
Invalid or unusual values
Cleaning & Validation Summary

Missing Values:
The datasets were checked for missing/null values using Python. Columns containing missing values were identified and reviewed before analysis.

Data Types:
Data types were inspected using info() and relevant columns were checked to ensure that numerical, categorical, and date fields were represented appropriately for analysis.

Duplicate Records:
Duplicate records were checked using duplicated(). Duplicate records, where identified, were handled to prevent them from affecting analysis and reporting.

Outlier Detection:
The IQR method was applied only to meaningful numerical/continuous variables rather than identifiers or code fields.

Key findings included:

Product Cost Price: 129 potential outliers were identified above the upper bound of 16,445.

Product Weight: Potential outliers were identified above the upper bound of 4,250 grams. During verification, 103 products had a repeated weight of 5,000 grams across different product types/categories, indicating a potential data-quality issue.

Return Amount: 1,045 potential outliers were identified above the upper bound of 171,769 The identified values ranged from approximately 171,791 to 1.13 million.

Supplier Payment Terms: This numerical field was considered appropriate for outlier analysis, while fields such as phone numbers, PIN codes, and IDs were excluded because they are identifiers/codes rather than measurable quantities.

Customer Age: The IQR check did not identify outliers requiring treatment.
Result

The cleaned and validated datasets are now prepared for the next phases of the IndiaCart project. The cleaning process focused on identifying data-quality issues while avoiding unnecessary removal of legitimate business values. The resulting data will be used for EDA, KPI calculation, Power BI dashboard development, and management insights.
